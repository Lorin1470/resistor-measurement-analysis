```python
# ==============================================================================
# 電子電路實習 - 實驗一：電阻量測誤差分析程式
# 說明：包含三組實驗電阻數據（色碼 10kΩ、色碼 2.2kΩ、精密 2.2kΩ）
# ==============================================================================

# 1. 安裝中文字型（若在 Google Colab 環境執行）
!apt-get install -y fonts-wqy-zenhei > /dev/null 2>&1

import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import numpy as np

# 設定 Matplotlib 中文字型，避免豆腐方塊
font_path = '/usr/share/fonts/truetype/wqy/wqy-zenhei.ttc'
fm.fontManager.addfont(font_path)
plt.rcParams['font.family'] = 'WenQuanYi Zen Hei'
plt.rcParams['axes.unicode_minus'] = False

# ==============================================================================
# 定義三組測量數據
# ==============================================================================
datasets = [
    {
        "title": "色碼電阻 10kΩ 誤差分析圖",
        "resistor_type": "色碼電阻 10kΩ",
        "unit": "kΩ",
        "true_value": 10.0,
        "measurements": np.array([9.885, 9.883, 9.882, 9.882, 9.881])
    },
    {
        "title": "色碼電阻 2.2kΩ 誤差分析圖",
        "resistor_type": "色碼電阻 2.2kΩ",
        "unit": "kΩ",
        "true_value": 2.2,
        "measurements": np.array([2.1746, 2.1739, 2.1734, 2.1739, 2.1734])
    },
    {
        "title": "精密電阻 2.2kΩ 誤差分析圖",
        "resistor_type": "精密電阻 2.2kΩ",
        "unit": "kΩ",
        "true_value": 2.2,
        "measurements": np.array([2.1850, 2.1851, 2.1851, 2.1853, 2.1854])
    }
]

# ==============================================================================
# 逐一計算誤差指標並繪製圖表
# ==============================================================================
for i, data in enumerate(datasets, start=1):
    measurements = data["measurements"]
    true_val = data["true_value"]
    unit = data["unit"]
    x_positions = np.arange(1, len(measurements) + 1)
    
    # 統計量計算
    mean_val = np.mean(measurements)
    systematic_error = mean_val - true_val
    std_dev = np.std(measurements, ddof=1)  # 樣本標準差（隨機誤差指標）
    pct_error = (abs(mean_val - true_val) / true_val) * 100
    
    # 終端機輸出報告用數據
    print(f"========== 【版本 {i}：{data['resistor_type']}】 ==========")
    print(f"1. 標稱真實值 (μ) : {true_val:.4f} {unit}")
    print(f"2. 測量平均值 (X̄) : {mean_val:.5f} {unit}")
    print(f"3. 固定系統偏差   : {systematic_error:+.5f} {unit}")
    print(f"4. 隨機誤差 (標準差 σ): {std_dev:.5f} {unit}")
    print(f"5. 百分比誤差     : {pct_error:.3f} %")
    print("-" * 50)
    
    # 繪圖
    plt.figure(figsize=(9, 5.5))
    
    # 真實值基準線（黑色實線）
    plt.axhline(y=true_val, color='black', linestyle='-', linewidth=2, 
                label=f'真實值 ({true_val} {unit})')
    
    # 隨機誤差範圍（藍色陰影區間）
    plt.axhspan(true_val - std_dev, true_val + std_dev, color='blue', alpha=0.2, 
                label=f'隨機誤差範圍 (±{std_dev:.4f} {unit})')
    
    # 測量值散布點（紅色圓點）
    plt.scatter(x_positions, measurements, color='red', marker='o', s=100, zorder=5, label='測量值')
    
    # 測量平均值線（紅色虛線）
    plt.axhline(y=mean_val, color='red', linestyle='--', linewidth=1.5, 
                label=f'測量平均值 ({mean_val:.4f} {unit})')
    
    # 動態自適應 Y 軸顯示範圍，防止圖形被邊界切斷
    y_min = min(np.min(measurements), true_val - std_dev)
    y_max = max(np.max(measurements), true_val + std_dev)
    padding = (y_max - y_min) * 0.4 if (y_max - y_min) > 0 else 0.01
    plt.ylim(y_min - padding, y_max + padding)
    
    # 圖表標註與美化
    plt.title(data["title"], fontsize=15, pad=12)
    plt.xlabel('測量次數', fontsize=12)
    plt.ylabel(f'測量值 ({unit})', fontsize=12)
    plt.xticks(x_positions)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend(loc='best', framealpha=0.9)
    plt.tight_layout()
    plt.show()
```
