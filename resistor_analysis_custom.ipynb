```python
# ==========================================
# 1. 安裝中文字型（Colab 環境）
# ==========================================
!apt-get install -y fonts-wqy-zenhei > /dev/null 2>&1

import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import numpy as np

# 直接掛載中文字型，免重啟 runtime 也不會噴豆腐方塊
font_path = '/usr/share/fonts/truetype/wqy/wqy-zenhei.ttc'
fm.fontManager.addfont(font_path)
plt.rcParams['font.family'] = 'WenQuanYi Zen Hei'
plt.rcParams['axes.unicode_minus'] = False  # 讓負號正常顯示

# ==========================================
# 2. 資料填寫區（換成你自己量到的數據！）
# ==========================================
# 把陣列裡面的數字改成你量測的數據（若是 5 次就填 5 個，10 次就填 10 個）
measurements = np.array([98.88, 97.2, 98.5, 98.3, 97.7, 98.4, 98.2, 98.1, 98.6, 97.9])
x_positions = np.arange(1, len(measurements) + 1)

# 填入你測量電阻的標稱值/真值（例如：100、2200 或 10000）
true_value = 100

# 隨機誤差範圍（講義範例固定寫 0.5；如果你要用實測標準差，也可以改寫成 np.std(measurements, ddof=1)）
random_error_range = 0.5 

# ==========================================
# 3. 計算平均值與系統誤差
# ==========================================
mean_value = np.mean(measurements)
systematic_error = mean_value - true_value

print(f"測量平均值: {mean_value:.3f} Ω")
print(f"系統偏差 (固定偏差): {systematic_error:.3f} Ω")

# ==========================================
# 4. 繪製誤差分析圖
# ==========================================
plt.figure(figsize=(10, 6))

# 繪製真實值線（黑色實線）
plt.axhline(y=true_value, color='black', linestyle='-', linewidth=2, label=f'真實值 ({true_value}Ω)')

# 繪製隨機誤差範圍（藍色陰影區間）
plt.axhspan(true_value - random_error_range, true_value + random_error_range,
            color='blue', alpha=0.2, label=f'隨機誤差範圍 (±{random_error_range}Ω)')

# 繪製實際量測點（紅色點）
plt.scatter(x_positions, measurements, color='red', marker='o', s=100, label='測量值')

# 繪製平均值線（紅色虛線）
plt.axhline(y=mean_value, color='red', linestyle='--', linewidth=1.5, label=f'測量平均值 ({mean_value:.2f}Ω)')

# 圖表外觀設定
plt.title('電阻測量誤差分析圖', fontsize=16)
plt.xlabel('測量次數', fontsize=12)
plt.ylabel('測量值 (Ω)', fontsize=12)
plt.xticks(x_positions)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(loc='lower left')

# 輸出圖表
plt.show()
```
